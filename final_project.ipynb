```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset.
# Ensure the dataset file exists and is in the same directory as this jupyter notebook.
HMS_df = pd.read_csv('HMS_2023-2024_qv RACE final.csv')

# Display the first few rows of the dataframe to ensure it is loaded correctly
display(HMS_df.head())

# Show columns and each column's data type
HMS_df.info() 
HMS_df.isnull().sum()
```

```python
# Question 1: What’s the Demographic makeup of my data?
# Create new variable 'data_by_race' to store the dataset
data_by_race = pd.DataFrame(HMS_df)

# Get the count value for each variables
race_counts = data_by_race['ALL_RACE'].value_counts()
sexbirth_count = data_by_race['sex_birth'].value_counts()
GenID_count = data_by_race['ALL_GENDER'].value_counts()

# Graph pie chart for racial breakdown of my data
race_counts.plot(kind='pie', title='Race Breakdown', autopct='%1.1f%%', pctdistance=1.2, labeldistance=None)
plt.legend(bbox_to_anchor=(1.0, 1.0),fontsize='small')
plt.ylabel('')
plt.show()

# Graph pie chart for sex at birth breakdown of my data
sexbirth_count.plot(kind='pie', title='Sex at Birth Breakdown', autopct='%1.1f%%', pctdistance=1.2, labeldistance=None)
plt.legend(loc="best")
plt.ylabel('')
plt.show()

# Graph pie chart for gender identity breakdown of my data
GenID_count.plot(kind='pie', title='Gender Identities Breakdown', autopct='%1.1f%%', pctdistance=1.2, labeldistance=None)
plt.legend(bbox_to_anchor=(1.0, 1.0),fontsize='small')
plt.ylabel('')
plt.show()

### need help with the number percentage on race breakdown
```

```python
# Question 2: Does financial status differ between racial groups for college students?
freq_table_FC = pd.crosstab(data_by_race['ALL_RACE'], data_by_race['fincur'])  
stacked_data_FC = freq_table_FC.apply(lambda x: x*100/sum(x), axis=1)
stacked_data_FC.plot(kind="barh", stacked=True)
plt.title('Current Financial Stress by Race')
plt.xlabel('Percentage')
plt.ylabel('Race')
plt.xticks(rotation=0)
plt.legend(title='Key Legend:',bbox_to_anchor=(1.0, 1.0),fontsize='small')
plt.show()

freq_table_FP = pd.crosstab(data_by_race['ALL_RACE'], data_by_race['finpast'])  
stacked_data_FP = freq_table_FP.apply(lambda x: x*100/sum(x), axis=1)
stacked_data_FP.plot(kind="barh", stacked=True)
plt.title('Past Financial Stress by Race')
plt.xlabel('Percentage')
plt.ylabel('Race')
plt.xticks(rotation=0)
plt.legend(title='Key Legend:',bbox_to_anchor=(1.0, 1.0),fontsize='small')
plt.show()

###how to move the ordering around
```
